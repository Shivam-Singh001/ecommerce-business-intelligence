```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for column in date_columns:
    orders[column] = pd.to_datetime(orders[column])
```

```python
orders.dtypes
```

```python
orders["delivery_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_purchase_timestamp"]
).dt.days
```

```python
orders["delivery_days"].describe()
```

```python
orders["delivery_delay_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_estimated_delivery_date"]
).dt.days
```

```python
orders["delivery_delay_days"].describe()
```

```python
print("Total orders:", len(orders))

print(
    "Orders with delivery date:",
    orders["order_delivered_customer_date"].notna().sum()
)

print(
    "Orders without delivery date:",
    orders["order_delivered_customer_date"].isna().sum()
)
```

```python
delivered_orders = orders[
    orders["order_delivered_customer_date"].notna()
].copy()

on_time_orders = delivered_orders[
    delivered_orders["delivery_delay_days"] <= 0
]

late_orders = delivered_orders[
    delivered_orders["delivery_delay_days"] > 0
]

print("Delivered orders:", len(delivered_orders))
print("On-time orders:", len(on_time_orders))
print("Late orders:", len(late_orders))

print(
    "On-time delivery rate:",
    round(len(on_time_orders) / len(delivered_orders) * 100, 2),
    "%"
)

print(
    "Late delivery rate:",
    round(len(late_orders) / len(delivered_orders) * 100, 2),
    "%"
)
```

# 2. Data Quality Assessment

Before performing business analysis, the datasets are checked for:

- Missing values
- Duplicate records
- Invalid or inconsistent values
- Incorrect data types
- Outliers
- Referential integrity between tables

The purpose is to ensure that subsequent analysis and business insights are based on reliable data.

```python
missing_summary = []

for name, data in datasets.items():
    for column in data.columns:
        missing_count = data[column].isna().sum()
        missing_pct = (missing_count / len(data)) * 100

        if missing_count > 0:
            missing_summary.append({
                "Dataset": name,
                "Column": column,
                "Missing Values": missing_count,
                "Missing %": round(missing_pct, 2)
            })

missing_df = pd.DataFrame(missing_summary)

missing_df.sort_values(
    "Missing %",
    ascending=False
)
```

```python
duplicate_summary = []

for name, data in datasets.items():
    duplicate_summary.append({
        "Dataset": name,
        "Duplicate Rows": data.duplicated().sum()
    })

duplicates_df = pd.DataFrame(duplicate_summary)

duplicates_df
```
